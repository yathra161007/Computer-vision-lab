```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = np.ones((500, 500, 3), dtype=np.uint8) * 255

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("White Image")
plt.axis("off")
plt.show()
```

```python
cv2.rectangle(image, (50, 50), (200, 200), (0, 0, 255), -1)
cv2.rectangle(image, (300, 50), (450, 200), (0, 255, 0), -1)
cv2.rectangle(image, (50, 300), (200, 450), (255, 0, 0), -1)
cv2.rectangle(image, (300, 300), (450, 450), (0, 255, 255), -1)
```

```python
plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("White Image with Four Colored Boxes")
plt.axis("off")
plt.show()
```
